## Q11. Predict the output. Why do the two matrices behave differently?

#### Expected output

```text
[[99, 0, 0], [0, 0, 0], [0, 0, 0]]
[[99, 0, 0], [99, 0, 0], [99, 0, 0]]
```

````python
matrix1 = [[0] * 3 for _ in range(3)]
matrix2 = [[0] * 3] * 3

matrix1[0][0] = 99
matrix2[0][0] = 99

print(matrix1)
print(matrix2)
````

#### Explanation

#### The comprehension creates three separate row lists, so only the first row of matrix1 changes. Multiplication repeats references to one row list, so all three rows of matrix2 show the change.

## Q12. Predict the output. Why do `x = x + [4]` and `a += [4]` behave differently?

#### Expected output

```text
x = [1, 2, 3, 4]
y = [1, 2, 3]
a = [1, 2, 3, 4]
b = [1, 2, 3, 4]
```

````python
x = [1, 2, 3]
y = x
x = x + [4]
print("x =", x)
print("y =", y)

a = [1, 2, 3]
b = a
a += [4]
print("a =", a)
print("b =", b)
````

#### Explanation

#### List addition creates a new list and reassigns x, leaving y pointing to the original list. For lists, += modifies the existing list in place, so both a and b show the added item.

## Q13. Predict the output. What is the value of `i` after the loop ends, and why?

#### Expected output

```text
4
NameError: name 'j' is not defined
```

````python
for i in range(5):
    pass
print(i)

for j in range(0):
    pass
print(j)  # Assumes j has not already been defined.
````

#### Explanation

#### The first loop leaves i equal to its last assigned value, 4; a for loop does not create a separate scope. The empty second loop never assigns j, so printing it raises NameError in a fresh session; a previously defined j would retain its old value.

## Q15. Predict the output. Explain Python’s short-circuit evaluation in one line.

#### Expected output

```text
5
5
hello
0
10
default
[0]
```

````python
print(0 or 5)
print(5 or 0)
print(None or "hello")
print(0 and 5)
print(5 and 10)
print([] or "default")
print([0] or "default")
````

#### Explanation

#### Python evaluates left to right and skips an unneeded operand: or returns the first truthy operand (otherwise the last), while and returns the first falsy operand (otherwise the last).

#### A non-empty list such as [0] is truthy even though its element is zero; an empty list is falsy.
