## Q31. Find all Armstrong numbers from 1 to 9999, print a comma-separated list and the count.

#### Answer

#### Count digits arithmetically, then sum each digit raised to that count.

```python
armstrong_numbers = []
for number in range(1, 10000):
    remaining = number
    digits = 0
    while remaining > 0:
        digits += 1
        remaining //= 10
    remaining = number
    total = 0
    while remaining > 0:
        total += (remaining % 10) ** digits
        remaining //= 10
    if total == number:
        armstrong_numbers.append(number)
for index in range(len(armstrong_numbers)):
    if index > 0:
        print(", ", end="")
    print(armstrong_numbers[index], end="")
print()
print("Count:", len(armstrong_numbers))
```

## Q33. Check primality using divisors only up to √N and print the sum of primes from 2 through N.

#### Answer

#### The condition divisor × divisor <= candidate avoids floating-point square roots and imports.

```python
n = int(input("N: "))
is_prime = n >= 2
divisor = 2
while divisor * divisor <= n and is_prime:
    if n % divisor == 0:
        is_prime = False
    divisor += 1
print("Prime:" if is_prime else "Not prime:", n)
prime_sum = 0
for candidate in range(2, n + 1):
    candidate_is_prime = True
    divisor = 2
    while divisor * divisor <= candidate:
        if candidate % divisor == 0:
            candidate_is_prime = False
            break
        divisor += 1
    if candidate_is_prime:
        prime_sum += candidate
print("Sum of primes:", prime_sum)
```

## Q35. Compute HCF with Euclid’s while-loop algorithm and obtain LCM from HCF × LCM = a × b.

```python
a = int(input("First positive integer: "))
b = int(input("Second positive integer: "))
if a <= 0 or b <= 0:
    print("Both integers must be positive.")
else:
    first, second = a, b
    while second != 0:
        first, second = second, first % second
    hcf = first
    lcm = (a // hcf) * b
    print("HCF:", hcf)
    print("LCM:", lcm)
```

## Q37. Parse ten student marks with split and map; without sorting, find maximum, minimum, average, count above average and second highest.

#### Second highest means second-highest distinct mark. If all marks are equal, no second distinct mark exists.

```python
try:
    marks = list(map(float, input("Enter 10 marks separated by spaces: ").split()))
    valid = len(marks) == 10
    for mark in marks:
        if not (0 <= mark <= 100):
            valid = False
    if not valid:
        print("Enter exactly 10 numeric marks from 0 to 100.")
    else:
        highest = lowest = marks[0]
        second_highest = None
        total = 0
        for mark in marks:
            total += mark
            if mark < lowest:
                lowest = mark
            if mark > highest:
                second_highest = highest
                highest = mark
            elif mark < highest and (second_highest is None or mark > second_highest):
                second_highest = mark
        average = total / len(marks)
        above_average = 0
        for mark in marks:
            if mark > average:
                above_average += 1
        print("Maximum:", highest)
        print("Minimum:", lowest)
        print(f"Average: {average:.2f}")
        print("Above average:", above_average)
        print("Second highest:", second_highest if second_highest is not None else "No second distinct mark")
except ValueError:
    print("All marks must be numeric.")
```

## Q38. Count review word frequencies case-insensitively with a dictionary; display descending frequencies and the most frequent word.

#### Answer

#### Alphanumeric words are separated by punctuation/whitespace. A manual selection loop avoids lambda; ties retain first-appearance order, including the reported most frequent word.

```python
review = input("Review: ")
cleaned = ""
for character in review.lower():
    if character.isalnum():
        cleaned += character
    else:
        cleaned += " "
frequencies = {}
for word in cleaned.split():
    frequencies[word] = frequencies.get(word, 0) + 1
remaining = dict(frequencies)
first_result = True
if not remaining:
    print("No words found.")
while remaining:
    best_word = None
    best_count = -1
    for word in remaining:
        if remaining[word] > best_count:
            best_word = word
            best_count = remaining[word]
    if first_result:
        most_frequent_word, most_frequent_count = best_word, best_count
        first_result = False
    print(f"{best_word}: {best_count}")
    del remaining[best_word]
if frequencies:
    print(f"Most frequent: {most_frequent_word} ({most_frequent_count})")
```

## Q45. Encrypt with a Caesar shift N, wrapping letters and preserving case; leave digits, spaces and special characters unchanged.

#### Answer

#### Shift English A–Z and a–z modulo 26. Negative or large shifts also work.

```python
message = input("Message: ")
shift = int(input("Shift: "))
encrypted = ""
for character in message:
    if "A" <= character <= "Z":
        encrypted += chr((ord(character) - ord("A") + shift) % 26 + ord("A"))
    elif "a" <= character <= "z":
        encrypted += chr((ord(character) - ord("a") + shift) % 26 + ord("a"))
    else:
        encrypted += character
print("Encrypted:", encrypted)
```

## Q46. Write calculate_gst(amount, gst_rate=18, gst_inclusive=False), returning rounded base, GST and total. Test all three given cases.

#### Answer

#### For inclusive GST, divide by 1 + rate/100 to recover the base. Returned floats are rounded to two decimals; formatting controls trailing zeros.

```python
def calculate_gst(amount, gst_rate=18, gst_inclusive=False):
    if amount < 0 or gst_rate < 0:
        raise ValueError("Amount and GST rate must be non-negative.")
    if gst_inclusive:
        total = amount
        base = amount / (1 + gst_rate / 100)
        gst = total - base
    else:
        base = amount
        gst = base * gst_rate / 100
        total = base + gst
    return round(base, 2), round(gst, 2), round(total, 2)

for result in (calculate_gst(1000), calculate_gst(1180, gst_inclusive=True),
               calculate_gst(500, gst_rate=5)):
    print(f"({result[0]:.2f}, {result[1]:.2f}, {result[2]:.2f})")
```

## Q47. Write recursive power(base, exp) without ** or pow, including zero and negative integer exponents.

#### Answer

#### Repeated squaring reduces recursion depth. A zero base with a negative exponent raises ZeroDivisionError; exponent zero returns 1 as specified.

```python
def power(base, exp):
    if not isinstance(exp, int):
        raise TypeError("Exponent must be an integer.")
    if exp == 0:
        return 1
    if exp < 0:
        if base == 0:
            raise ZeroDivisionError("Zero cannot have a negative exponent.")
        return 1 / power(base, -exp)
    half = power(base, exp // 2)
    if exp % 2 == 0:
        return half * half
    return base * half * half

print("power(2, 10):", power(2, 10))
print("power(5, 0):", power(5, 0))
print("power(3, -2):", power(3, -2))
```

#### Answer

#### Percentage growth uses the earlier day as denominator; slicing aligns the six consecutive pairs.

```python
import numpy as np
sales = np.array([12000, 18500, 14200, 22100, 19800, 35200, 28400])
days = np.array(["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"])
average = np.mean(sales)
daily_changes = (sales[1:] - sales[:-1]) / sales[:-1] * 100
print("Total weekly sales: ₹", np.sum(sales))
print(f"Daily average: ₹{average:.2f}")
print("Maximum sales day:", days[np.argmax(sales)])
print("Days above average:", np.count_nonzero(sales > average))
print(f"Monday–Sunday growth: {(sales[-1] - sales[0]) / sales[0] * 100:.2f}%")
print("Day-on-day changes (%):", np.round(daily_changes, 2))
```

## Q50. Create a 5×5 NumPy array with border 1 and interior 0, without loops or comprehensions.

#### Answer

#### Start with ones and overwrite the interior using a two-dimensional slice.

```python
import numpy as np
border = np.ones((5, 5), dtype=int)
border[1:-1, 1:-1] = 0
print(border)
```
