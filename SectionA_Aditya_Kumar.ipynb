## Q1. What is the difference between the `is` operator and the `==` operator? Give one example where two variables return True for `==` but False for `is`.

#### `==` compares values, while `is` checks whether two names refer to the same object in memory.

#### For `a = [1, 2]` and `b = [1, 2]`, `a == b` is `True`, but `a is b` is `False` because the lists are separate objects.

## Q2. Classify the following as Mutable or Immutable: list, tuple, set, dict, str, int, float, bool, frozenset. Explain in one line why immutability matters in real code.

#### Mutable: `list`, `set`, `dict`.

#### Immutable: `tuple`, `str`, `int`, `float`, `bool`, `frozenset`.

#### Immutability prevents in-place changes to an object, reducing unintended changes when the object is shared.

## Q3. Predict the output of `print(0.1 + 0.2 == 0.3)`. Then explain why this happens.

#### Output: `False`.

#### Binary floating-point cannot represent these decimal fractions exactly; `0.1 + 0.2` evaluates to `0.30000000000000004`, which differs from `0.3`.

#### A tolerance check such as `abs((0.1 + 0.2) - 0.3) < 1e-9` returns `True`.

## Q4. Both list and tuple can store ordered collections. Give 3 concrete situations where you would choose a tuple over a list, and explain the reasoning behind each.

#### Fixed coordinates: `(latitude, longitude)` prevents replacing either item in place.

#### Dictionary keys: `(year, month)` is hashable when both elements are integers, whereas a list cannot be a dictionary key.

#### Fixed records: `(student_id, name)` expresses a fixed structure and prevents adding, removing or replacing its fields in place.

## Q5. Why is dictionary lookup O(1) on average while searching for a value in a list using `in` is O(n)? Explain in plain English without using the word "hash" — then explain it again with the word "hash".

#### Without the technical term: a dictionary uses a key to calculate where to look, usually requiring only a few checks; a list may require checking every item.

#### With the technical term: a dictionary uses a hash table for O(1) average key lookup, while list membership uses an O(n) sequential search.

#### This comparison concerns dictionary keys; searching dictionary values is generally O(n).

## Q6. What is the difference between `range()` and `numpy.arange()`? Mention at least 3 differences, one involving floats.

#### `range()` is built into Python and returns a compact sequence object; `numpy.arange()` requires NumPy and allocates an array of values.

#### `range()` accepts integer arguments; `numpy.arange()` also accepts floating-point steps, which can introduce rounding effects.

#### NumPy arrays support element-wise arithmetic, while `range` objects do not support those array operations.

## Q7. What is the difference between `list.append(x)`, `list.extend(x)`, and `list.insert(i, x)`? For each, show what happens when `x = [4, 5]` is added to `a = [1, 2, 3]`.

#### Use a fresh `a = [1, 2, 3]` for each example, with `x = [4, 5]`.

#### `a.append(x)` adds one item at the end: `[1, 2, 3, [4, 5]]`.

#### `a.extend(x)` adds each element of x: `[1, 2, 3, 4, 5]`.

#### `a.insert(1, x)` adds x as one item at index 1: `[1, [4, 5], 2, 3]`.

## Q8. What does `a, *b, c = [10, 20, 30, 40, 50]` do, and what are the values of a, b, c? What happens with `[10, 20]` and `[10]`?

#### Starred unpacking assigns the first item to a, the last to c, and collects the middle items into list b.

#### For `[10, 20, 30, 40, 50]`: `a = 10`, `b = [20, 30, 40]`, `c = 50`.

#### For `[10, 20]`: `a = 10`, `b = []`, `c = 20`.

#### For `[10]`: `ValueError`, because there are not enough items for both a and c.

## Q9. Explain the difference between `/`, `//`, `%`, and `**` operators in Python. For each, give the output when applied to (-7, 2).

#### `/` performs true division: `-7 / 2` gives `-3.5`.

#### `//` performs floor division, rounding toward negative infinity: `-7 // 2` gives `-4`.

#### `%` gives the remainder consistent with floor division: `-7 % 2` gives `1`, since `-7 = (-4 × 2) + 1`.

#### `**` performs exponentiation: `(-7) ** 2` gives `49`. Parentheses matter: `-7 ** 2` would give `-49`.

## Q10. Why is NumPy significantly faster than Python lists for numerical operations? Give 3 reasons, at least one mentioning memory layout.

#### For standard numeric dtypes, NumPy stores values in compact, homogeneous buffers; contiguous arrays improve cache use and memory efficiency.

#### Vectorized operations use compiled loops, avoiding the interpreter overhead of a Python loop over individual elements.

#### A fixed numeric dtype reduces per-element type handling, and many operations can use SIMD instructions to process multiple values together.
